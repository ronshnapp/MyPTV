```python
from myptv.particle_matching_mod import matching_with_marching_particles_algorithm
from myptv.io_mod import read_from_file, write_to_file

import numpy as np
import pandas as pd
```

```python
from myptv.imaging_mod import camera_wrapper, img_system

camNames = ['cam1','cam2','cam3']
cams = [camera_wrapper(cn,'.') for cn in camNames]
for c in cams:
    c.load()

imsys = img_system(cams)
blob_files = ['blobs_cam1.hdf5', 'blobs_cam2.hdf5', 'blobs_cam3.hdf5']

ROI= (0.0, 70.0, 0.0, 70.0, -50.0, 15.0)
voxel_size = 20
N0 = 0
max_d_err = 0.25
min_cam_match = 3
```

```python
mp = matching_with_marching_particles_algorithm(imsys, blob_files, max_d_err, ROI, N0, 
                                                 voxel_size, min_cam_match=3, reverse_eta_zeta=True)

mp.get_blobs()
```

```python
savename = 'test_matching.hdf5'

frames = sorted(list(mp.frames))

frm_res = mp.match_frame(frames[0], savename, False)

for frm in frames[1:]:
    frm_res = mp.match_frame(frm, savename, True)
```
